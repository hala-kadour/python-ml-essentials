# Working with Lists in Python

A **list** is an ordered, mutable collection of items. Lists can store elements of different data types and allow duplicate values.

In this notebook, we will cover:
1. Creating lists and accessing items by **index**.
2. **Slicing** lists to extract specific portions.
3. Adding elements (`append`, `insert`, `extend`).
4. Removing elements (`pop`, `remove`).

## 1. Creating Lists and Accessing Items
Lists are defined using square brackets `[]`. Each item in a list has a zero-based **index**.

```python
# Creating a list of programming languages
languages = ["Python", "Dart", "JavaScript", "SQL"]

# Accessing elements using positive and negative indices
print("First item:", languages[0])       # Python
print("Last item:", languages[-1])       # SQL
print("Second to last:", languages[-2])  # JavaScript
```

## 2. List Slicing
Slicing allows you to extract a sub-section of a list using the syntax `list[start:stop]`.

```python
# Sample dataset
frameworks = ["Flutter", "React", "Vue", "Angular", "Django"]

# Extracting a range of items
print("First three frameworks:", frameworks[:3])   # Indices 0, 1, 2
print("Frameworks from index 2:", frameworks[2:])  # Index 2 to end
print("Middle slice (1 to 3):", frameworks[1:4])   # Indices 1, 2, 3
```

## 3. Adding Data to Lists
Python provides multiple methods to insert new items into an existing list:
* **`append()`**: Adds a single item to the end of the list.
* **`insert()`**: Adds an item at a specific index.
* **`extend()`**: Merges another list/iterable into the current list.

```python
# Initialize an empty list
task_queue = []

# Appending single elements
task_queue.append("Configure database")
task_queue.append("Write unit tests")

# Inserting at a specific position (index 0 = top priority)
task_queue.insert(0, "Fix high-priority bug")

# Extending with multiple elements
new_tasks = ["Review PR", "Deploy to staging"]
task_queue.extend(new_tasks)

print("Updated Task Queue:", task_queue)
```

## 4. Removing Data from Lists
You can remove items by value using `remove()` or by index using `pop()`.

```python
# Removing by value using remove()
task_queue.remove("Write unit tests")

# Retrieving and removing by index using pop()
completed_task = task_queue.pop(0)  # Removes the first item

print(f"Completed Task: '{completed_task}'")
print("Remaining Tasks:", task_queue)
```
