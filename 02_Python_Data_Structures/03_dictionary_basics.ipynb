# Dictionary Basics in Python

A **dictionary** is an unordered collection of key-value pairs[cite: 13]. Dictionaries are optimized for retrieving data quickly using keys rather than numeric indices[cite: 9, 13].

In this notebook, we will cover:
1. Creating dictionaries in different ways.
2. Adding and updating key-value pairs[cite: 8].
3. Safe data retrieval techniques to prevent `KeyError` exceptions.

## 1. Creating Dictionaries
You can create dictionaries using curly braces `{}` or the built-in `dict()` function[cite: 11, 13]. Keys must be unique and immutable (e.g., strings or numbers)[cite: 13].

```python
# Creating a dictionary with literal syntax
app_settings = {
    "theme": "dark",
    "language": "en",
    "notifications_enabled": True
}

# Creating an empty dictionary using dict()
user_profile = dict()

# Creating a dictionary from keyword arguments
server_info = dict(host="localhost", port=5432)

print(app_settings)
print(server_info)
```

## 2. Adding and Updating Data
Assigning a value to a key will insert it if the key does not exist, or overwrite it if it already exists[cite: 8].

```python
# Adding new key-value pairs
user_profile["username"] = "hala_dev"
user_profile["role"] = "admin"

print("Initial Profile:", user_profile)

# Updating an existing key
user_profile["role"] = "super_admin"

print("Updated Profile:", user_profile)
```

## 3. Safe Data Retrieval
Accessing a missing key using standard bracket notation `dict["key"]` raises a `KeyError`[cite: 6]. Python provides safer ways to handle missing keys[cite: 6]:
* **`try` / `except KeyError`**: Catches missing keys explicitly[cite: 6].
* **`.get(key, default)`**: Returns `None` or a specified fallback value if the key is missing[cite: 6].

```python
student_record = {"name": "Hiba", "gpa": 3.8}

# 1. Unsafe retrieval (Raises KeyError if uncommented)
# score = student_record["score"]

# 2. Handling KeyError with try/except
try:
    score = student_record["score"]
except KeyError:
    print("Key 'score' not found in record.")

# 3. Safe retrieval using .get() with a default fallback
score = student_record.get("score", 0.0)
print(f"Student Score: {score}")

# Removing items safely using pop()
removed_gpa = student_record.pop("gpa", None)
print(f"Removed GPA: {removed_gpa}, Remaining Record: {student_record}")
```
