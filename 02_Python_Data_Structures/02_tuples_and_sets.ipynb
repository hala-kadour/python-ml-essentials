# Working with Tuples and Sets in Python

Besides lists, Python offers two other essential sequence data structures:
1. **Tuples (`tuple`)**: Immutable (read-only) ordered collections.
2. **Sets (`set`)**: Unordered collections of unique elements.

In this notebook, we will explore how to use both effectively and understand their best use cases.

## 1. Tuples (`tuple`)
Tuples are created using parentheses `()`. They behave like lists, but their contents **cannot be modified** after creation (immutable)[cite: 9].

### Why use Tuples?
* **Data Integrity**: Protects data from accidental modifications[cite: 9].
* **Performance**: Slightly faster and lightweight compared to lists.

```python
# Defining a configuration tuple (read-only)
server_config = ("127.0.0.1", 8080, "HTTPS")

# Accessing elements using index
print("Host IP:", server_config[0])
print("Port Number:", server_config[1])

# Unpacking a tuple into variables
host, port, protocol = server_config
print(f"Connecting to {host}:{port} via {protocol}")
```

### Tuple Immutability
Attempting to add, modify, or delete elements from a tuple will trigger a `TypeError`[cite: 9].

```python
# Demonstrating immutability (uncommenting below causes a TypeError)
# server_config[1] = 9090  # TypeError: 'tuple' object does not support item assignment

# Checking available methods (Tuples only have 'count' and 'index')
print("Index of 'HTTPS':", server_config.index("HTTPS"))
```

## 2. Sets (`set`)
Sets are created using curly braces `{}` or the `set()` constructor[cite: 9]. Sets automatically remove duplicates and are ideal for membership testing[cite: 9].

```python
# Creating a set with duplicate values
user_ids = {101, 102, 103, 101, 104, 102}

# Duplicates are automatically removed
print("Unique User IDs:", user_ids)  # Output: {101, 102, 103, 104}

# Creating an empty set (Must use set(), not {} which creates a dict)
active_sessions = set()
active_sessions.add("session_abc")
active_sessions.add("session_abc")
active_sessions.add("session_xyz")
print("Active Sessions:", active_sessions)
```

### Fast Membership Testing
Checking if an item exists in a set is extremely fast compared to a list.

```python
allowed_roles = {"admin", "editor", "moderator"}

# Checking existence using 'in'
current_role = "editor"

if current_role in allowed_roles:
    print(f"Access granted for role: {current_role}")
else:
    print("Access denied.")
```
