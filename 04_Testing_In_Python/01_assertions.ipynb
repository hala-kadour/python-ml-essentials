# Assertions in Python

An **assertion** is a statement that something must be true at a specific point in a program. If the condition is false, Python raises an `AssertionError`. Assertions are most useful for catching mistakes close to where they happen.

In this notebook, we will cover:
1. Writing basic assertions with helpful error messages.
2. Using assertions to validate internal assumptions.
3. Distinguishing assertions from exceptions intended for users.

## 1. Basic Assertions

The `assert` statement takes a condition and an optional error message:

`assert condition, "message"`

If the condition evaluates to `False`, execution stops with an `AssertionError` that includes the message.

```python
def normalize_score(score):
    assert 0 <= score <= 100, f"Score must be between 0 and 100, got {score}"
    return score / 100

normalized = normalize_score(85)
print("Normalized score:", normalized)
```

When an assertion fails, the message should explain which assumption was violated and include the value that caused the problem. This makes failures much faster to diagnose.

```python
try:
    normalize_score(140)
except AssertionError as error:
    print("Assertion failed:", error)
```

## 2. Validating Internal Assumptions

Assertions are a good fit for assumptions that must hold for your own code to be correct, such as valid split sizes or consistent data shapes. They are not a replacement for handling bad input from users or external systems.

```python
def split_training_data(total_rows, train_fraction=0.8):
    assert total_rows > 0, "total_rows must be positive"
    assert 0 < train_fraction < 1, "train_fraction must be between 0 and 1"

    train_rows = int(total_rows * train_fraction)
    validation_rows = total_rows - train_rows
    return train_rows, validation_rows

train_rows, validation_rows = split_training_data(10_000, 0.8)
print(f"Training rows: {train_rows:,}")
print(f"Validation rows: {validation_rows:,}")
```

## 3. Assertions vs. Exceptions

Use assertions to document programmer assumptions during development and testing. Use explicit exceptions such as `ValueError` for invalid input that may come from outside your program.

Assertions are also disabled when Python runs with the `-O` flag, so they must never be used to enforce security or business rules.

```python
def validate_learning_rate(rate):
    if rate <= 0:
        raise ValueError("Learning rate must be greater than zero")
    return rate

try:
    validate_learning_rate(0)
except ValueError as error:
    print("Input error:", error)

print("Valid rate:", validate_learning_rate(0.01))
```
