# Exception Handling in Python

An **exception** is an error that occurs during the execution of a program. If not handled, an exception will cause the program to crash.

In this notebook, we will cover:
1. **How Exceptions Happen**: Raising built-in and custom errors.
2. **`try` / `except` Blocks**: Catching specific errors to prevent crashes.
3. **Catching Multiple Exceptions**: Handling different error types.
4. **Capturing Error Messages**: Accessing the exception's details.

## 1. How Exceptions Happen
Python raises errors when performing invalid operations (e.g., dividing by zero). You can also manually raise an exception using the `raise` keyword.

```python
# Unhandled exception example (commented out to prevent execution failure)
# total_score = 100 / 0  # Raises ZeroDivisionError

# Manually raising a custom exception
invalid_age = -5

if invalid_age < 0:
    raise ValueError("Age cannot be negative!")
```

## 2. Catching Exceptions with `try` / `except`
Instead of allowing the program to crash, use `try` and `except` blocks to handle errors gracefully.

> **Best Practice:** Always catch specific exceptions (e.g., `ZeroDivisionError`) instead of using bare `except:` or generic `except Exception:`.

```python
# Handling a division error safely
total_users = 0
total_clicks = 150

try:
    average_clicks = total_clicks / total_users
except ZeroDivisionError:
    print("Warning: Cannot divide by zero. Setting average clicks to 0.")
    average_clicks = 0

print(f"Average Clicks: {average_clicks}")
```

## 3. Catching Multiple Exceptions
A single block of code can trigger different types of exceptions. You can chain multiple `except` blocks to handle each error type individually.

```python
# Handling type mismatches and missing keys
user_data = {"name": "Alex", "age": "twenty"}

try:
    # Attempting to calculate birth year
    birth_year = 2026 - user_data["age"]  # Type error
    
except KeyError:
    print("Error: The requested key does not exist in the dictionary.")
except TypeError:
    print("Error: Invalid data type for calculation. 'age' must be an integer.")
```
