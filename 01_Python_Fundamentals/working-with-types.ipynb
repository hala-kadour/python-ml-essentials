# Working with Data Types in Python

In Python, every value has a specific **data type**. Data types define what kind of operations you can perform on a variable. 

In this notebook, we will explore five essential Python data types:
1. **Strings (`str`)** - Text data
2. **Integers (`int`)** - Whole numbers
3. **Floats (`float`)** - Numbers with decimals
4. **Booleans (`bool`)** - Logical values (`True` or `False`)
5. **NoneType (`None`)** - Represents empty or missing values

## 1. Strings (`str`)
Strings represent text. You can define strings using single (`'`), double (`"`), or triple (`"""`) quotes.

```python
# strings can be assigned with single, double, and triple quotes
user_name = 'Adam'
course_title = "Python for Machine Learning"
multi_line_text = """This is a multi-line string.
It is useful for long explanations or documentation."""

print(user_name)
print(course_title)
print(multi_line_text)
```

```python
# Triple quotes are useful when having quotes within a string
summary = """Triple quotes are useful when you have a ' or a " within a string"""
print(summary)
```

```python
# Use single quotes when you have double quotes in a string
summary = 'Use single quotes when there is a " in your string'
print(summary)
print("And use double quotes when you have a ' in your string")
```

```python
# string support adding to other strings to compose text
greeting = "Hello " + user_name
print(greeting)
```

### String Formatting with f-strings
F-strings provide a clean and modern way to insert variables directly into strings.

```python
# Combining variables using f-strings
score = 95
greeting = f"Hello {user_name}, your test score is {score}%!"
print(greeting)
```

## 2. Numbers: Integers (`int`) & Floats (`float`)
* **`int`**: Represents whole numbers (e.g., `10`, `-5`).
* **`float`**: Represents numbers with decimal points (e.g., `10.5`, `3.14`).

```python
# Checking variable types using type()
items_count = 5
unit_price = 12.99

print(type(items_count))  # Output: <class 'int'>
print(type(unit_price))   # Output: <class 'float'>
```

### Mathematical Operations & Type Conversion
Be careful when performing arithmetic operations. Division (`/`) always converts the result into a `float`.

```python
# Division always returns a float
total_amount = 100
split_parts = 4
result = total_amount / split_parts

print(result)        # Output: 25.0
print(type(result))  # Output: <class 'float'>

# Converting string to integer (Type Casting)
str_number = "50"
converted_number = int(str_number)
print(converted_number + 10)  # Output: 60
```

```python
# watchout for invalid mathematical operations
7 / 0
```

```python
# as with all other types, watch out when mixing unsupported types
7 + "14"
```

## 3. Booleans (`bool`)
Booleans represent one of two values: `True` or `False`. They are essential for logical evaluations and controlling program execution.

```python
type(True)
```

```python
# Boolean evaluations
is_logged_in = True
has_access = False

print(type(is_logged_in))

# Truthy and Falsy conversions using bool()
# Numbers other than 0 evaluate to True. Empty collections or 0 evaluate to False.
print(bool(1))       # Output: True
print(bool(0))       # Output: False
print(bool(""))      # Output: False (empty string)
print(bool("Data"))  # Output: True
```

## 4. NoneType (`None`)
`None` is used to signal the absence of a value or a null value. It is often returned by functions that don't explicitly return anything.

```python
# Assigning None to a variable
selected_option = None

print(type(selected_option))  # Output: <class 'NoneType'>

# Best practice: Check for None using 'is'
if selected_option is None:
    print("No option has been selected yet.")
```
