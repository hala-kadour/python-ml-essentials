# Logic and Conditionals in Python

Conditional statements allow your program to make decisions and execute specific blocks of code based on certain conditions.

In this notebook, we will cover:
1. **`if` Statements**: Executing code when a condition is `True`.
2. **Truthy and Falsy Values**: How Python evaluates non-boolean data types.
3. **`else` and `elif`**: Handling multiple alternative outcomes.
4. **Logical Operators**: Combining conditions using `and`, `or`, and `not`.

## 1. Basic `if` Statements
An `if` statement evaluates an expression. If the expression evaluates to `True`, the indented code block executes.

```python
# Simple conditional check
user_role = "admin"

if user_role == "admin":
    print("Access granted: Welcome to the dashboard!")
```

## 2. Truthy and Falsy Evaluations
In Python, values other than explicit `True` and `False` can evaluate to booleans in conditional statements:
* **Falsy values**: `0`, `""` (empty string), `[]` (empty list), `None`.
* **Truthy values**: Any non-zero number, non-empty strings, and populated collections.

```python
# Checking empty vs populated data structures
cart_items = []

# Empty list evaluates to False
if cart_items:
    print("Processing your order...")

# Adding an item makes the list Truthy
cart_items.append("Laptop")

if cart_items:
    print(f"You have {len(cart_items)} item(s) in your cart.")
```

## 3. Alternative Branches: `else` and `elif`
* Use **`else`** to execute a fallback block when the `if` condition fails.
* Use **`elif`** (short for *else if*) to test multiple distinct conditions sequentially.

```python
# Evaluating user access level
account_status = "suspended"

if account_status == "active":
    print("User can place orders.")
elif account_status == "pending":
    print("User needs to verify their email.")
else:
    print("Account is restricted. Please contact support.")
```

## 4. Logical Operators (`and`, `or`, `not`)
* **`and`**: Returns `True` only if **all** conditions are met.
* **`or`**: Returns `True` if **at least one** condition is met.
* **`not`**: Reverses the evaluation result.

```python
# Combining multiple conditions using 'and' and 'not'
is_email_verified = True
has_valid_payment = True
is_blocked = False

# Checking eligibility for checkout
if is_email_verified and has_valid_payment and not is_blocked:
    print("Checkout successful!")
else:
    print("Unable to complete checkout.")
```
