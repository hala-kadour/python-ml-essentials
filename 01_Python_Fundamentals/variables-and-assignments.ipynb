# Variables and assignments

Variable is a named location im memory that stores a value, while assignment is when you give a variable a value by using the = operator.


```python
# assigning variables
animal = "Panda"
```

```python
#python doesn't require explicit type declartion because it is a dynamiclly-typed langauge (String, int, float)
home_town = "china"
type = "Giant Panda"
age = 2
weight = 44.4
```

```python
# re-assigning variables
type = "Red Panda"
type
```

```python
# using print() with variables
print(animal, home_town)
```

```python
# Create new variables from existing variables
# watch out with strings
# full_name = name + last_name
animal_info = f"Animal : {animal}, Type : {type}, Home Town : {home_town}"
animal_info
```

```python
# the print() function helps display values
# it adds separation when using commas
print(animal, "is", age, "years old.")
```
