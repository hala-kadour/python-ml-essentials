# Functions Fundamentals in Python

A **function** is a block of organized, reusable code used to perform a single, related action. Functions help make code modular, readable, and maintainable.

In this notebook, we will cover:
1. Defining basic functions and returning values.
2. Documenting functions using **Docstrings**.
3. Exploring built-in documentation inspection (`__doc__` and `help()`).

## 1. Defining Functions & Returning Values
Functions are defined using the `def` keyword followed by the function name and parentheses `()`. Use `return` to send back a value to the caller.

```python
# A simple function returning a string
def get_favorite_framework():
    return "Flutter"

# Executing the function and saving the result
framework = get_favorite_framework()
print("Favorite Framework:", framework)
```

## 2. Documenting Functions with Docstrings
A **Docstring** is a string literal placed as the first statement inside a function. It explains what the function does, its parameters, and what it returns.

```python
def calculate_tax(amount, rate=0.15):
    """Calculates total tax for a given amount and rate.

    Parameters:
    amount (float): The base cost.
    rate (float): The tax percentage (default is 0.15).

    Returns:
    float: Total tax amount.
    """
    return amount * rate

# Testing the function
tax = calculate_tax(100.0)
print("Tax amount:", tax)
```

## 3. Reading Function Documentation
Python provides ways to inspect docstrings programmatically using the `__doc__` attribute or the built-in `help()` function.

```python
# Accessing docstring via __doc__
print("Docstring via __doc__:")
print(calculate_tax.__doc__)

# Inspecting function via help()
print("-" * 40)
help(calculate_tax)
```
