# Function Arguments, Scope, and Closures

Python offers flexible ways to pass arguments to functions and handle variables across different scopes.

In this notebook, we will cover:
1. **Positional** vs. **Keyword** arguments with default values.
2. Dynamic arguments using `*args` and `**kwargs`.
3. Nested functions, **variable scope**, and `nonlocal`.
4. Creating **Closures** to retain state.

## 1. Positional and Keyword Arguments
* **Positional Arguments**: Assigned based on their order.
* **Keyword Arguments**: Assigned by parameter name, allowing custom ordering and default values.

```python
def configure_database(host="localhost", port=5432, ssl_enabled=True):
    print(f"Connecting to {host}:{port} (SSL: {ssl_enabled})")

# Using default values
configure_database()

# Overriding specific values using keyword arguments
configure_database(port=3306, host="db.example.com")
```

## 2. Dynamic Arguments (*args & **kwargs)
* **`*args`**: Accepts any number of extra positional arguments as a **tuple**.
* **`**kwargs`**: Accepts any number of extra keyword arguments as a **dictionary**.

```python
# Function using both *args and **kwargs
def build_query(table, *columns, **filters):
    cols = ", ".join(columns) if columns else "*"
    query = f"SELECT {cols} FROM {table}"
    
    if filters:
        conditions = [f"{k} = '{v}'" for k, v in filters.items()]
        query += " WHERE " + " AND ".join(conditions)
        
    return query

# Testing dynamic query building
sql_1 = build_query("users", "id", "name", role="admin", active="true")
sql_2 = build_query("products")

print("Query 1:", sql_1)
print("Query 2:", sql_2)
```

## 3. Nested Functions and Scope (`nonlocal`)
A function defined inside another function can access variables from the outer scope. To modify an outer non-global variable, we use the `nonlocal` keyword.

```python
def create_counter():
    count = 0  # Outer scope variable
    
    def increment():
        nonlocal count  # Allows modifying the outer 'count' variable
        count += 1
        return count
        
    return increment

# Instantiating the counter function
counter = create_counter()
print("Count 1:", counter())
print("Count 2:", counter())
```

## 4. Retaining State with Closures
A **Closure** is an inner function that retains access to variables from its parent scope even after the parent function has finished executing.

```python
def make_multiplier(factor):
    # 'factor' is saved inside the closure
    def multiply(number):
        return number * factor
    return multiply

# Creating specialized multiplier functions
double = make_multiplier(2)
triple = make_multiplier(3)

print("Double 5:", double(5))
print("Triple 5:", triple(5))
```
