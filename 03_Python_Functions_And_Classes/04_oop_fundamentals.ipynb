# Object-Oriented Programming (OOP) - Fundamentals

Object-Oriented Programming (OOP) allows you to structure code into reusable **Classes** and **Objects**, binding data and functionality together.

In this notebook, we will cover:
1. Defining a **Class** and instantiating **Objects**.
2. The `__init__` constructor and the `self` keyword.
3. **Instance Variables** vs. **Class Variables**.
4. Writing custom instance methods.

## 1. Class Definition and Instantiation
A **Class** acts as a blueprint for creating objects. An **Object** is a concrete instance created from that class.

```python
# Defining a simple class
class UserProfile:
    pass

# Instantiating objects
user1 = UserProfile()
user2 = UserProfile()

print("User 1 object:", user1)
print("Are they the same object?", user1 is user2)
```

## 2. Constructor (`__init__`) and `self`
* **`__init__`**: A special method called automatically when creating a new instance to initialize attributes.
* **`self`**: Refers to the current instance of the class, allowing access to its properties and methods.

```python
class DatabaseConnection:
    def __init__(self, db_name, port=5432):
        # Setting instance attributes
        self.db_name = db_name
        self.port = port
        self.is_connected = False

# Creating instances with initial values
dev_db = DatabaseConnection("dev_db")
prod_db = DatabaseConnection("prod_db", port=5433)

print(f"Dev DB: {dev_db.db_name} on port {dev_db.port}")
print(f"Prod DB: {prod_db.db_name} on port {prod_db.port}")
```

## 3. Instance Variables vs. Class Variables
* **Instance Variables**: Unique to each specific instance (`self.attribute`).
* **Class Variables**: Shared across all instances of the class.

```python
class AppConfig:
    # Class Variable (Shared across all instances)
    app_version = "1.0.0"
    
    def __init__(self, environment):
        # Instance Variable (Unique to each object)
        self.environment = environment

dev_config = AppConfig("development")
prod_config = AppConfig("production")

print(f"Dev Env: {dev_config.environment} | Version: {dev_config.app_version}")
print(f"Prod Env: {prod_config.environment} | Version: {prod_config.app_version}")

# Modifying class variable affects all instances
AppConfig.app_version = "1.1.0"
print("Updated Version for Dev:", dev_config.app_version)
```

## 4. Custom Instance Methods
Methods are functions defined inside a class that operate on the instance attributes using `self`.

```python
class ShoppingCart:
    def __init__(self, customer_name):
        self.customer_name = customer_name
        self.items = []

    def add_item(self, item_name, price):
        self.items.append({"name": item_name, "price": price})
        print(f"Added '{item_name}' to {self.customer_name}'s cart.")

    def get_total(self):
        total = sum(item["price"] for item in self.items)
        return total

# Using cart methods
cart = ShoppingCart("Hala")
cart.add_item("Keyboard", 45)
cart.add_item("Mouse", 20)

print("Total Price:", cart.get_total())
```
