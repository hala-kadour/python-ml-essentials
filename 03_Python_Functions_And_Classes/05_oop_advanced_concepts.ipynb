## 1. Inheritance and `super()`
Inheritance allows a **child class** to reuse and extend attributes/methods from a **parent class**. The `super()` function calls methods from the parent class directly.

```python
# Base / Parent Class
class BaseService:
    def __init__(self, service_name):
        self.service_name = service_name

    def log_status(self):
        print(f"[{self.service_name}] Service is running.")

# Child Class inheriting from BaseService
class EmailService(BaseService):
    def __init__(self, service_name, provider):
        super().__init__(service_name)  # Initialize parent attributes
        self.provider = provider

    # Method Overriding
    def log_status(self):
        print(f"[{self.service_name}] Connected via {self.provider}.")

# Testing inheritance
email_bot = EmailService("AuthNotifier", "SendGrid")
email_bot.log_status()
```

## 2. Multiple Inheritance
Python supports inheriting from multiple parent classes simultaneously, giving the child class access to all parent methods.

```python
class LoggerMixin:
    def log(self, message):
        print(f"[LOG]: {message}")

class ValidatorMixin:
    def validate_email(self, email):
        return "@" in email and "." in email

# Child Class inheriting from multiple mixins
class UserRegistration(LoggerMixin, ValidatorMixin):
    def register(self, email):
        if self.validate_email(email):
            self.log(f"User '{email}' registered successfully.")
        else:
            self.log(f"Failed: Invalid email address '{email}'.")

app = UserRegistration()
app.register("user@example.com")
app.register("invalid-email")
```

## 3. Special Dunder Methods (Magic Methods)
Dunder (Double Underscore) methods allow custom classes to interface with built-in Python behaviors:
* **`__str__`**: Human-readable string representation.
* **`__len__`**: Defines behavior for `len()`.
* **`__call__`**: Makes instances callable like regular functions.

```python
class CustomPipeline:
    def __init__(self, name, steps):
        self.name = name
        self.steps = steps

    def __str__(self):
        return f"Pipeline(Name: {self.name}, Total Steps: {len(self.steps)})"

    def __len__(self):
        return len(self.steps)

    def __call__(self, input_data):
        print(f"Executing pipeline '{self.name}' on data: {input_data}")

# Instantiating custom pipeline
pipeline = CustomPipeline("DataCleaner", ["strip", "lowercase", "tokenize"])

print(str(pipeline))         # Invokes __str__
print("Steps count:", len(pipeline)) # Invokes __len__
pipeline("Raw User Input")    # Invokes __call__
```

## 4. Encapsulation with `@property`
The `@property` decorator converts a class method into a read-only getter attribute, protecting private data members (`_variable`).

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance  # Protected attribute

    @property
    def balance(self):
        """Read-only balance property."""
        return f"${self._balance:,.2f}"

    def deposit(self, amount):
        if amount > 0:
            self._balance += amount

# Usage
account = BankAccount("Hala", 1500)
print("Account Balance:", account.balance)

# Modifying through method works
account.deposit(500)
print("Updated Balance:", account.balance)

# Attempting direct reassignment (account.balance = 3000) will raise an AttributeError!
```

## 5. Class Methods and Static Methods
* **`@classmethod`**: Binds to the class (`cls`) rather than the instance. Useful for alternative constructors.
* **`@staticmethod`**: Utility methods that do not access class (`cls`) or instance (`self`) state.

```python
class DateFormatter:
    def __init__(self, year, month, day):
        self.year = year
        self.month = month
        self.day = day

    @classmethod
    def from_string(cls, date_str):
        """Alternative constructor: parses 'YYYY-MM-DD'."""
        year, month, day = map(int, date_str.split("-"))
        return cls(year, month, day)

    @staticmethod
    def is_valid_year(year):
        return 1900 <= year <= 2100

# Using alternative constructor
date1 = DateFormatter.from_string("2026-07-28")
print(f"Parsed Date: {date1.year}/{date1.month}/{date1.day}")

# Using static method directly from class
print("Is 2026 valid?", DateFormatter.is_valid_year(2026))
```
